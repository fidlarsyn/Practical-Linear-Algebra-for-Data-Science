# **Set Vektor (Vector Sets)**
Set vektor adalah sekumpulan vektor yang dikelompokkan bersama, biasanya disimbolkan dengan huruf kapital miring (seperti S atau V). Set ini bisa berisi jumlah vektor yang terbatas (seperti dataset COVID-19 dari berbagai negara) atau jumlah yang tak terbatas (seperti pada subspace).

# **Kombinasi Linear Berbobot (Linear Weighted Combination)**
Ini adalah operasi fundamental yang mencampur informasi dari beberapa variabel dengan memberikan bobot (skalar) yang berbeda pada setiap vektor. Secara teknis, ini hanyalah operasi perkalian skalar-vektor yang diikuti oleh penjumlahan vektor.


**Contoh Kode (Kombinasi Linear):**

```python
import numpy as np

# Menentukan bobot (skalar)
l1, l2, l3 = 1, 2, -3

# Menentukan set vektor
v1 = np.array([5-7])
v2 = np.array([-4, 0, -4])
v3 = np.array([7-9])

# Menghitung kombinasi linear berbobot
w = l1*v1 + l2*v2 + l3*v3
# Hasil: array([-7, -4, -13])
```

# **Kemandirian Linear (Linear Independence)**
Kemandirian linear adalah sifat dari sebuah set vektor. Sebuah set dikatakan terikat linear (dependent) jika setidaknya satu vektor dalam set tersebut dapat dinyatakan sebagai kombinasi linear dari vektor lainnya (vektor redundan). Sebaliknya, set dikatakan mandiri linear (independent) jika tidak ada vektor yang bisa dibentuk dari kombinasi vektor lainnya. Secara matematis, kemandirian linear diuji dengan melihat apakah ada kombinasi bobot non-nol yang dapat menghasilkan vektor nol (0).

# **Subspace dan Span**
* **Subspace**: Merupakan set vektor tak terhingga yang dihasilkan dari semua kemungkinan kombinasi linear dari sebuah set vektor asal.
* **Span**: Adalah mekanisme atau "kata kerja" untuk menciptakan subspace tersebut.Misalnya, satu vektor tunggal akan melakukan span pada sebuah garis (subspace 1D), sedangkan dua vektor yang mandiri linear akan melakukan span pada sebuah bidang datar (subspace 2D) dalam ruang yang lebih tinggi.


**Contoh Kode (Membuat Titik dalam Subspace):**

```python
# Membuat 100 titik acak yang berada dalam span satu vektor [7, 8]
# Ini akan membentuk garis lurus di plot
scalars = np.random.uniform(-4, 4, 100)
v = np.array([7, 8])

# Menggunakan outer product untuk mendapatkan koordinat semua titik
points = np.outer(scalars, v)
```

# **Basis**
Basis adalah "penggaris" yang digunakan untuk mengukur dan mendeskripsikan informasi dalam sebuah ruang. Sebuah set vektor dapat menjadi basis untuk sebuah subspace jika memenuhi dua syarat mutlak:
1. Set tersebut harus melakukan span pada subspace tersebut.
2. Set tersebut harus mandiri secara linear.

Kemandirian linear sangat penting karena menjamin bahwa setiap titik dalam ruang tersebut memiliki koordinat yang unik (hanya satu cara kombinasi linear untuk mencapai titik tersebut).
