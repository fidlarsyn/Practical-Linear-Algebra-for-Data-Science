# **PCA (Principal Component Analysis) menggunakan Eigendecomposition dan SVD**
Tujuan utama PCA adalah menemukan set basis vektor baru (komponen utama) yang searah dengan variansi maksimal dalam data. PCA membantu dalam reduksi dimensi dengan mempertahankan fitur paling penting dan membuang derau.

Langkah-langkah Implementasi PCA:
1. Mean-center setiap fitur dalam data.
2. Hitung matriks kovarians fitur-oleh-fitur.
3. Lakukan eigendecomposition pada matriks kovarians tersebut.
4. Urutkan eigenvalues secara menurun dan sesuaikan eigenvectors-nya.
5. Konversi eigenvalues menjadi persentase varians untuk interpretasi.

**Kumpulan Kode (Logika PCA):**

```python
# Menghitung PCA melalui eigendecomposition matriks kovarians
# X_centered adalah data yang sudah dikurangi rata-ratanya
cov_mat = (X_centered.T @ X_centered) / (n_samples - 1)
evals, evecs = np.linalg.eig(cov_mat)

# Mengurutkan komponen
indices = np.argsort(evals)[::-1]
evals = evals[indices]
evecs = evecs[:, indices]
```

# **Linear Discriminant Analysis (LDA)**
Berbeda dengan PCA yang memaksimalkan varians secara umum, LDA adalah teknik klasifikasi yang mencari arah untuk memaksimalkan pemisahan antar kategori data.
Logika Matematika: LDA bekerja dengan memaksimalkan rasio varians between-category terhadap varians within-category.

Solusi: Implementasi LDA menggunakan generalized eigendecomposition pada dua matriks kovarians (matriks CB untuk antar-kelas dan CW untuk dalam-kelas).

Contoh Konsep:
Eigenvectors dari hasil ini memberikan bobot fitur yang paling baik dalam memisahkan kelas. LDA akan selalu mencari solusi linear, bahkan jika data tidak dapat dipisahkan secara linear secara sempurna.
