# **Matriks Ortogonal (Orthogonal Matrices)**
Matriks ortogonal adalah matriks persegi di mana kolom-kolomnya saling ortogonal (tegak lurus satu sama lain) dan masing-masing memiliki magnitudo (norma) tepat 1.0. Properti unik dari matriks ini adalah hasil kali transpose dengan dirinya sendiri menghasilkan matriks identitas (Q^T Q=I), yang berarti inversinya sangat mudah dihitung karena Q^−1=Q^T. Matriks ini penting karena dapat memutar vektor tanpa mengubah panjangnya.

**Contoh Kode (Properti Matriks Ortogonal):**

```python
import numpy as np

# Mengasumsikan Q adalah matriks ortogonal dari hasil QR
# Properti utama: Q.T @ Q = I
print(np.round(Q.T @ Q, 4))

# Properti inversi: Q.T == Q_inv
Q_inv = np.linalg.inv(Q)
# Selisih antara transpose dan inversi mendekati nol
print(np.linalg.norm(Q.T - Q_inv))
```

# **Gram-Schmidt**
Prosedur Gram-Schmidt adalah metode untuk mengubah matriks non-ortogonal menjadi matriks ortogonal. Algoritmenya bekerja secara iteratif pada kolom matriks dari kiri ke kanan: kolom saat ini di-ortogonalisasi terhadap semua kolom sebelumnya menggunakan dekomposisi vektor ortogonal, lalu hasilnya dinormalisasi menjadi unit vektor. Meskipun secara konsep sangat penting untuk memahami pembentukan matriks Q, prosedur ini jarang digunakan dalam praktik karena masalah ketidakstabilan numerik.

# **Dekomposisi QR (QR Decomposition)**
Dekomposisi ini memecah matriks A menjadi dua bagian: Q (matriks ortogonal) dan R (matriks segitiga atas/upper-triangular). Matriks R menyimpan informasi yang "hilang" selama proses ortogonalisasi kolom-kolom A menjadi Q. Matriks R selalu berbentuk segitiga atas karena proses ortogonalisasi kolom yang dilakukan secara berurutan dari kiri ke kanan.

**Contoh Kode (Implementasi QR):**

```python
A = np.random.randn(6, 6)

# Melakukan dekomposisi QR
Q, R = np.linalg.qr(A)

# Memverifikasi A = QR
print(np.linalg.norm(A - Q @ R)) # Hasil sangat kecil mendekati 0
```

# **Ukuran Q dan R (Sizes of Q and R)**
Ukuran matriks hasil dekomposisi bergantung pada apakah kita menggunakan metode "economy" (juga disebut reduced) atau "full" (complete).
* **Economy QR**: Untuk matriks tinggi (M>N), Q akan berukuran M×N.
* **Full QR**: Q akan berbentuk matriks persegi M×M, di mana kolom-kolom tambahannya dibuat dari vektor-vektor ortogonal yang dikonstruksi secara matematis untuk melengkapi ruang tersebut.

# **QR dan Inversi (QR and Inverses)**
Dekomposisi QR menyediakan cara yang lebih stabil secara numerik untuk menghitung inversi matriks dibandingkan algoritme tradisional. Dengan formula A=QR, maka inversinya dapat dihitung sebagai A^−1 =R^−1 Q^T. Menggunakan Q^T jauh lebih stabil daripada menghitung inversi matriks padat secara langsung, dan menginversi matriks segitiga R juga lebih efisien melalui prosedur back substitution.

**Contoh Kode (Inversi via QR):**

```python
# A_inv = inv(R) @ Q.T
A_inv_qr = np.linalg.inv(R) @ Q.T

# Bandingkan dengan fungsi inversi standar
A_inv_std = np.linalg.inv(A)
print(np.linalg.norm(A_inv_qr - A_inv_std))
```
