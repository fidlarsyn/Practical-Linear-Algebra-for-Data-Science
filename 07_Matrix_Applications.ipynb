# **Multivariate Data Covariance Matrices**

Pemfilteran gambar adalah proses deteksi fitur di mana sebuah kernel (template kecil) dicocokkan dengan bagian-bagian gambar melalui operasi konvolusi. Secara teknis, ini melibatkan perkalian Hadamard dan penjumlahan elemen dalam jendela geser. Salah satu aplikasi populernya adalah penghalusan gambar (smoothing) menggunakan kernel Gaussian.

**Contoh Kode (Membuat Kernel Gaussian 2D):**
