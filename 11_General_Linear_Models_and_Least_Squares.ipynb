# **General Linear Models**
GLM adalah kerangka kerja statistik yang menghubungkan variabel prediktor (independen) dengan variabel observasi (dependen). Dalam statistik, kita menggunakan terminologi yang sedikit berbeda dari aljabar linear murni:
* **Matriks Desain (X)**: Matriks di mana kolom adalah variabel independen.
* **Koefisien Regresi (β)**: Vektor bobot yang ingin kita cari nilainya.
* **Variabel Dependen (y)**: Vektor hasil observasi atau data. Persamaan standarnya dalam aljabar linear adalah Ax=b, yang dalam statistik ditulis sebagai Xβ=y.

Menyusun Model Linear (Setting Up a GLM)
Menyusun model melibatkan pembuatan matriks desain X. Salah satu komponen krusial adalah intercept (konstanta), yaitu kolom berisi angka 1 yang memungkinkan garis model tidak harus melewati titik asal (0,0).

**Contoh Kode (Menyusun Matriks Desain dengan Intercept):**

```python
import numpy as np

# Data contoh: jumlah kursus yang diambil
numcourses = [4, 12, 3, 14, ...]

# Menyusun matriks desain X (kolom 1: intercept, kolom 2: data)
X = np.hstack((np.ones((20,1)), np.array(numcourses, ndmin=2).T))
```

# **Menyelesaikan GLM: Solusi Least Squares**
Karena data di dunia nyata jarang sekali berada tepat di ruang kolom matriks desain (y∈/C(X)), kita mencari solusi yang meminimalkan jarak antara prediksi model dan data asli. Solusi ini disebut Least Squares (Kuadrat Terkecil) yang menggunakan left-inverse.
Rumus Utama: β=(X^T X)^−1 X^T y.

**Contoh Kode (Solusi Analitik Least Squares):**

```python
# Menghitung left-inverse secara manual untuk mencari beta
X_leftinv = np.linalg.inv(X.T @ X) @ X.T
beta = X_leftinv @ y
```

Perspektif Geometris dan **Residual**
Secara geometris, solusi least squares adalah proyeksi ortogonal dari vektor data y ke dalam subspace yang dibentuk oleh kolom-kolom matriks desain X. Jarak antara data asli y dan hasil proyeksi Xβ disebut sebagai residual atau error (ϵ).
* **Sifat Penting**: Vektor residual ϵ selalu tegak lurus (ortogonal) terhadap ruang kolom matriks desain X.

# Solusi via Dekomposisi QR
Menghitung (X^T X)^−1 secara langsung dapat menyebabkan ketidakstabilan numerik jika matriks memiliki condition number yang tinggi. Dekomposisi QR menyediakan cara yang lebih stabil untuk menyelesaikan masalah least squares tanpa perlu mengalikan X^T X secara eksplisit.

**Contoh Kode (Least Squares via QR):**

```python
# Melakukan dekomposisi QR pada X
Q, R = np.linalg.qr(X)

# Solusi beta: R^-1 @ Q^T @ y
# (Dalam prakteknya lebih stabil menggunakan back-substitution pada R)
beta_qr = np.linalg.inv(R) @ Q.T @ y
```
