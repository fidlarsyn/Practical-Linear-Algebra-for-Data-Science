# **Membuat dan Memvisualisasikan Vektor di NumPy**
Secara aljabar, vektor adalah daftar angka yang berurutan. Karakteristik penting dari vektor adalah dimensionalitas (jumlah elemen di dalamnya) dan orientasi (apakah vektor tersebut berbentuk baris/tidur atau kolom/berdiri). Dalam Python, orientasi sangat krusial karena beberapa operasi akan menghasilkan kesalahan jika orientasinya salah.

**Contoh Kode (Membuat Vektor):**

```python
import numpy as np

# Membuat vektor baris (1D array)
v_row = np.array([4-6])

# Membuat vektor kolom (2D array dengan reshape)
v_col = np.array([[4-6]]).T
```

# **Geometri Vektor**
Secara geometris, vektor dipahami sebagai garis lurus yang memiliki panjang (magnitude) dan arah (sudut). Vektor biasanya digambarkan dalam "posisi standar", di mana pangkalnya berada di titik asal (0,0) dan ujungnya berada di koordinat yang sesuai dengan elemen angka di dalamnya.

# **Operasi pada Vektor**
* **Penjumlahan**: Dilakukan dengan menjumlahkan setiap elemen yang bersesuaian. Syarat utamanya adalah kedua vektor harus memiliki dimensionalitas dan orientasi yang sama.
* **Perkalian Skalar**: Mengalikan setiap elemen vektor dengan sebuah angka tunggal (skalar). Secara geometris, ini mengubah panjang vektor tanpa mengubah arahnya (kecuali jika dikalikan angka negatif, maka arahnya berbalik 180 derajat).
* **Transpose**: Operasi untuk mengubah vektor kolom menjadi vektor baris, atau sebaliknya.

**Contoh Kode (Operasi Dasar):**

```python
v = np.array([4-6])
w = np.array([16-18])
s = 2

# Penjumlahan
v_plus_w = v + w

# Perkalian Skalar
v_scaled = s * v

# Transpose (menggunakan method .T)
v_transpose = v.T
```

# **Magnitude Vektor dan Vektor Unit**
Magnitude (panjang geometris) dihitung menggunakan rumus jarak Euclidean: akar kuadrat dari jumlah kuadrat elemen-elemennya. Vektor Unit adalah vektor yang memiliki panjang tepat 1.0 namun tetap mempertahankan arah yang sama dengan vektor asalnya.

**Contoh Kode (Magnitude):**

```python
# Menghitung magnitude (norm) menggunakan NumPy
mag = np.linalg.norm(v)

# Membuat vektor unit
v_unit = v / np.linalg.norm(v)
```

# **Perkalian Titik (Vector Dot Product)**
Dot product adalah salah satu operasi paling penting dalam aljabar linear. Hasilnya adalah satu angka skalar tunggal yang diperoleh dari jumlah hasil perkalian elemen-elemen yang bersesuaian. Operasi ini merupakan ukuran kemiripan (similarity) antar dua variabel dalam data science.

**Contoh Kode (Dot Product):**

```python
v = np.array([4, 26-28])
w = np.array([5, 6, 29, 30])

# Menghitung dot product
dp = np.dot(v, w) # Hasil: 70
```

# **Perkalian Vektor Lainnya**
* **Hadamard Multiplication**: Perkalian elemen-demi-elemen yang menghasilkan vektor baru dengan dimensi yang sama.
* **Outer Product**: Perkalian antara vektor kolom dan vektor baris yang menghasilkan sebuah matriks.

**Contoh Kode (Hadamard & Outer):**

```python
# Hadamard (element-wise) menggunakan tanda bintang
v_hadamard = v * w

# Outer product
v_outer = np.outer(v, w)
```
