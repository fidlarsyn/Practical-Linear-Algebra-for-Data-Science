# **Multikolinearitas (Multicollinearity)**
Dalam aljabar linear, multikolinearitas hanyalah istilah lain untuk ketergantungan linear antar kolom dalam matriks desain. Jika kolom-kolomnya terikat secara linear, matriks menjadi reduced-rank (singular) dan tidak memiliki left-inverse, sehingga solusi least squares tidak unik dan tidak stabil secara numerik.

# **Regularisasi (Regularization)**
Regularisasi adalah teknik untuk meningkatkan stabilitas numerik dan kemampuan generalisasi model dengan cara "menggeser" (shifting) matriks desain. Metode yang umum digunakan adalah Ridge Regression (regularisasi L^2), yang menambahkan nilai kecil pada diagonal matriks X^T X.
Rumus Utama: β=(X^T X+γ∥X∥F^2 I)^−1 X^T y.

**Contoh Kode (Least Squares dengan Regularisasi):**

```python
gamma = 0.01 # Parameter regularisasi
# Menghitung beta dengan menambahkan proporsi norma Frobenius pada diagonal
XTX = X.T @ X
reg_term = gamma * np.linalg.norm(X, 'fro')**2 * np.eye(X.shape[13])
beta_reg = np.linalg.inv(XTX + reg_term) @ X.T @ y
```
