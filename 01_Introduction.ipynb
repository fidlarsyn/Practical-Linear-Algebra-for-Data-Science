**Kumpulan Kode (Intuisi Komutativitas)**:

Cuplikan ini menunjukkan bagaimana kode digunakan untuk menguji prinsip komutativitas (a×b=b×a) secara empiris

```python
import numpy as np

# Menghasilkan dua angka acak
a = np.random.randn()
b = np.random.randn()

# Jika hasilnya 0.0, maka urutan perkalian tidak berpengaruh (komutatif)
print(a*b - b*a)
```
