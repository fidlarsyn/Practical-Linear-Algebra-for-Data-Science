# **Gambaran Besar SVD (The Big Picture of the SVD)**
SVD mendekomposisi matriks A (ukuran M×N, rank r) menjadi perkalian tiga matriks: A=UΣV^T.
* **U (M×M)**: Matriks ortogonal berisi vektor singular kiri. Kolom-kolomnya merupakan basis ortonormal untuk ruang kolom C(A).
* **Σ(M×N)**: Matriks diagonal berisi nilai singular (σ). Nilai-nilai ini selalu non-negatif dan diurutkan dari yang terbesar ke terkecil.
**V^T (N×N)**: Matriks ortogonal berisi vektor singular kanan. Baris-barisnya (atau kolom-kolom V) merupakan basis ortonormal untuk ruang baris R(A).

# **Nilai Singular dan Rank Matriks**
Rank sebuah matriks didefinisikan sebagai jumlah nilai singular yang tidak nol. Dalam komputasi praktis, ambang batas (tolerance) sering digunakan untuk membedakan nilai singular kecil yang sebenarnya adalah nol namun terkena kesalahan pembulatan komputer.

# **SVD dalam Python**
Di Python, SVD diimplementasikan melalui fungsi np.linalg.svd(). Perlu diperhatikan bahwa fungsi ini mengembalikan nilai singular dalam bentuk vektor (bukan matriks diagonal) dan mengembalikan V^T (bukan V).

**Contoh Kode (Implementasi SVD):**

```python
import numpy as np

# Membuat matriks acak nonsquare
A = np.random.randn(5, 3)

# Menghitung SVD
U, s, Vt = np.linalg.svd(A)

# Mengubah vektor s menjadi matriks diagonal Sigma yang sesuai ukuran A
Sigma = np.zeros(A.shape)
np.fill_diagonal(Sigma, s)

# Verifikasi rekonstruksi: A = U @ Sigma @ Vt
print(np.allclose(A, U @ Sigma @ Vt)) # True
```

# **SVD dan "Lapisan" Peringkat-1 (SVD and Rank-1 Layers)**
Matriks A dapat dipandang sebagai jumlah dari "lapisan-lapisan" matriks peringkat-1. Setiap lapisan dibentuk oleh perkalian luar (outer product) antara satu vektor singular kiri, satu nilai singular, dan satu vektor singular kanan. Low-rank approximation adalah teknik merekonstruksi matriks hanya menggunakan k lapisan pertama (di mana k<r), yang sering kali cukup untuk menangkap informasi utama sambil membuang derau.

**Contoh Kode (Aproksimasi Peringkat Rendah):**

```python
# Merekonstruksi A hanya dengan 2 lapisan pertama
k = 2
Ak = U[:, :k] @ np.diag(s[:k]) @ Vt[:k, :]
```

# Hubungan SVD dengan Eigendecomposition
Secara konsep, SVD dapat diturunkan dari eigendecomposition matriks simetris yang dihasilkan dari A:
* Vektor singular kiri (U) adalah vektor eigen dari AA^T.
* Vektor singular kanan (V) adalah vektor eigen dari A^T A.
* Nilai singular (σ) adalah akar kuadrat dari nilai eigen (λ) yang tidak nol dari AA^T atau A^T A.

# **Condition Number (Nomor Kondisi)**
