# **Norma Matriks dan Trace (Matrix Norms and Trace)**
Norma matriks adalah angka tunggal yang merincikan "ukuran" atau magnitudo elemen dalam matriks. Norma yang paling umum adalah Norma Frobenius (norma ℓ2), yang dihitung sebagai akar kuadrat dari jumlah kuadrat seluruh elemen matriks. Sementara itu, Trace adalah jumlah dari elemen-elemen pada diagonal utama matriks persegi.


**Contoh Kode (Norma Frobenius & Trace):**

```python
import numpy as np

A = np.array([[8-10], [11-13], [14-16]])

# 1. Menghitung Norma Frobenius
norm_F = np.linalg.norm(A, 'fro')

# 2. Menghitung Trace
tr_A = np.trace(A)

# 3. Hubungan Norma Frobenius dengan Trace: ||A|| = sqrt(tr(A.T @ A))
norm_via_trace = np.sqrt(np.trace(A.T @ A))
```

# **Ruang Matriks: Kolom, Baris, dan Null (Matrix Spaces)**
Matriks dapat dipandang sebagai kumpulan vektor yang membentuk ruang geometris:
* **Column Space (C(A))**: Seluruh kombinasi linear yang mungkin dari kolom-kolom matriks.
* **Row Space (R(A))**: Seluruh kombinasi linear dari baris-baris matriks.
* **Null Space (N(A))**: Set vektor y yang jika dikalikan dengan matriks A akan menghasilkan vektor nol (Ay=0).

**Contoh Kode (Mencari Null Space):**

```python
from scipy.linalg import null_space

# Mencari basis untuk Null Space matriks A
ns = null_space(A)
```

# **Rank Matriks**
Rank adalah angka yang menunjukkan jumlah maksimum kolom (atau baris) yang bersifat mandiri secara linear dalam sebuah matriks. Jika rank sama dengan jumlah kolom/baris terkecil, matriks disebut full-rank; jika lebih kecil, disebut reduced-rank atau singular. Konsep ini penting untuk mengetahui apakah suatu sistem persamaan memiliki solusi.

**Contoh Kode (Menghitung Rank):**

```python
# Menghitung rank matriks menggunakan NumPy
r = np.linalg.matrix_rank(A)
```

# **Determinan dan Karakteristik Polinomial**
Determinan adalah angka unik yang hanya dimiliki matriks persegi. Nilai determinan nol menandakan matriks tersebut adalah reduced-rank (singular) dan tidak dapat dibalik. Karakteristik Polinomial adalah persamaan yang menggabungkan pergeseran matriks (shifting) dengan determinan, yang merupakan kunci untuk menemukan Eigenvalues di bab selanjutnya.


**Contoh Kode (Determinan):**

```python
# Menghitung determinan matriks persegi
d = np.linalg.det(A)
```
