# **Sistem Persamaan (Systems of Equations)**
Langkah pertama dalam menyelesaikan masalah dunia nyata adalah mengubah sekumpulan persamaan menjadi persamaan matriks Ax=b. Di sini, A adalah matriks koefisien, x adalah vektor variabel yang tidak diketahui, dan b adalah vektor konstanta. Operasi pada persamaan ini harus dilakukan secara konsisten pada kedua sisi.

**Contoh Kode (Menyelesaikan Persamaan Matriks AX=B):**

```python
import numpy as np

# A dan B adalah matriks acak
A = np.random.randn(4,4)
B = np.random.randn(4,4)

# Mencari X dengan mengalikan B dengan invers A di sisi kiri
# Urutan perkalian sangat penting (tidak komutatif)
X = np.linalg.inv(A) @ B
```

# **Eliminasi Baris dan Echelon Form (Row Reduction)**
Eliminasi baris adalah proses menerapkan operasi perkalian skalar dan penjumlahan antar baris untuk mengubah matriks menjadi bentuk echelon (segitiga atas). Secara matematis, proses ini dapat dinyatakan sebagai L^−1 A=U, di mana L^−1 menyimpan semua transformasi linear yang dilakukan selama eliminasi.

# **Eliminasi Gaussian dan Back Substitution**
Eliminasi Gaussian mengubah matriks augmented menjadi bentuk echelon. Setelah variabel terakhir ditemukan, kita menggunakan back substitution (substitusi balik) untuk menghitung nilai variabel lainnya secara berurutan. Meskipun jarang dilakukan secara manual oleh ilmuwan data, konsep ini penting untuk memahami efisiensi algoritme.

# **Eliminasi Gauss-Jordan dan RREF**
Ini adalah versi lanjutan di mana setiap pivot (angka pertama non-nol di tiap baris) diubah menjadi 1, dan semua angka di atas pivot tersebut dihilangkan. Hasil akhirnya disebut Reduced Row Echelon Form (RREF). Matriks identitas biasanya akan muncul di bagian kiri matriks RREF jika matriks asalnya memiliki peringkat penuh (full-rank).

# **Dekomposisi LU (LU Decomposition)**
Dekomposisi LU memecah matriks A menjadi dua matriks segitiga: L (Lower/Segitiga Bawah) dan U (Upper/Segitiga Atas). Dalam implementasi praktis, matriks ketiga yang disebut Permutation Matrix (P) ditambahkan untuk menangani pertukaran baris, sehingga persamaannya menjadi A=PLU.

**Contoh Kode (Implementasi LU dengan SciPy):**

```python
from scipy.linalg import lu

# A adalah matriks persegi acak
A = np.random.randn(4,4)

# P = Permutasi, L = Segitiga Bawah, U = Segitiga Atas
P, L, U = lu(A)

# Verifikasi: A = P @ L @ U
print(np.linalg.norm(A - P @ L @ U)) # Hasil mendekati 0
```
