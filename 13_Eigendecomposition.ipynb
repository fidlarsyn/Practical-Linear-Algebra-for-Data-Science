# **Interpretasi Nilai Eigen dan Vektor Eigen**
* **Geometri**: Vektor eigen adalah vektor khusus yang arahnya tidak berubah (tidak berotasi) saat dikalikan dengan matriks; ia hanya meregang atau menyusut sebesar faktor yang disebut nilai eigen.
* **Statistik (PCA)**: Dalam analisis komponen utama, vektor eigen menunjukkan arah kovarians maksimal dalam data, sementara nilai eigen menunjukkan besarnya varians di arah tersebut.
* **Reduksi Derau & Kompresi**: Data dapat dibersihkan atau dikompresi dengan membuang dimensi yang memiliki nilai eigen sangat kecil, dengan asumsi nilai tersebut hanya mewakili derau (noise).

# **Mencari Nilai Eigen (Finding Eigenvalues)**
Nilai eigen (λ) ditemukan melalui polinomial karakteristik, yaitu dengan mengasumsikan matriks yang digeser (A−λI) adalah matriks singular (memiliki determinan nol). Secara matematis, kita mencari solusi λ yang memenuhi persamaan det(A−λI)=0.

# **Mencari Vektor Eigen (Finding Eigenvectors)**
Setelah nilai eigen ditemukan, vektor eigen yang bersesuaian diperoleh dengan mencari basis untuk null space dari matriks yang telah digeser oleh nilai eigen tersebut (A−λI). Penting untuk dicatat bahwa vektor eigen hanya penting dalam hal arah, bukan panjangnya; sehingga setiap versi yang diskalakan dari vektor eigen tetap merupakan vektor eigen yang valid.

**Contoh Kode (Eigendecomposition Standar):**

```python
import numpy as np

# Membuat matriks persegi
A = np.array([[15, 16]])

# Mencari nilai dan vektor eigen menggunakan NumPy
evals, evecs = np.linalg.eig(A)

print(f"Nilai Eigen: {evals}")
print(f"Vektor Eigen (Kolom): \n{evecs}")
```

# **Diagonalisasi Matriks Persegi (Diagonalizing a Square Matrix)**
Diagonalisasi adalah proses mewakili matriks A dalam bentuk A=VΛV^−1, di mana V adalah matriks yang kolom-kolomnya adalah vektor eigen, dan Λ (lambda kapital) adalah matriks diagonal yang berisi nilai-nilai eigen. Ini adalah cara untuk mengekspresikan informasi matriks dalam basis yang lebih sederhana dan "jarang" (sparse).

**Contoh Kode (Rekonstruksi Matriks):**

```python
# Membuat matriks diagonal nilai eigen
L = np.diag(evals)

# Matriks vektor eigen
V = evecs

# Rekonstruksi A = V @ L @ inv(V)
A_recon = V @ L @ np.linalg.inv(V)
```

# **Bentuk Kuadratik, Definiteness, dan Nilai Eigen**
Bentuk kuadratik (w^T Aw) menghasilkan skalar yang tandanya ditentukan oleh tanda nilai eigen matriks tersebut. Matriks disebut positive definite jika semua nilai eigennya positif, yang menjamin adanya solusi minimum yang unik dalam masalah optimasi seperti gradient descent. Matriks kovarians data selalu bersifat positive (semi)definite.
