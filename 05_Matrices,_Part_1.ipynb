# **Membuat, Mengindeks, dan Melakukan Slicing Matriks**
Dalam NumPy, matriks dapat dibuat dengan mengatur ulang (reshape) array atau menumpuk vektor-vektor. Karena matriks bisa sangat besar, visualisasi sering dilakukan menggunakan gambar (heatmap) daripada mencetak angka. Pengindeksan bersifat 0-based (dimulai dari nol), dan slicing memungkinkan kita mengambil sub-bagian tertentu dari matriks.


**Contoh Kode (Membuat & Slicing Matriks):**

```python
import numpy as np

# Membuat matriks 6x10 dan mengubah bentuknya
A = np.arange(60).reshape(6,10)

# Mengambil sub-matriks (baris 2-4 dan kolom 1-5)
# Menggunakan pola [start:stop:step]
sub = A[1:4, 0:5]
```

# **Matriks Khusus (Special Matrices)**
Terdapat beberapa kategori matriks yang sering digunakan karena propertinya yang unik:
Random: Matriks berisi angka acak (seringkali distribusi Gaussian).
* **Square vs Nonsquare**: Matriks persegi (jumlah baris = kolom) atau persegi panjang.
* **Diagonal**: Matriks yang hanya memiliki nilai non-nol pada garis diagonal utamanya.
* **Identity (I)**: Matriks diagonal persegi dengan semua nilai diagonalnya adalah 1.
* **Zeros**: Matriks yang semua elemennya adalah nol.

**Contoh Kode (Membuat Matriks Khusus):**

```python
# Matriks acak Gaussian 4x6
A = np.random.randn(4,6)

# Matriks Identitas 4x4
I = np.eye(4)

# Matriks Diagonal dari sebuah list
D = np.diag([9-12])
```

# **Operasi Matriks: Penjumlahan, Skalar, dan Hadamard**
Operasi ini bersifat intuitif karena dilakukan elemen-demi-elemen (element-wise):
* **Penjumlahan/Pengurangan**: Hanya berlaku untuk matriks dengan ukuran yang sama.
* **Perkalian Skalar**: Mengalikan setiap elemen dengan sebuah angka tunggal.
* **Perkalian Hadamard**: Perkalian elemen-demi-elemen antara dua matriks dengan ukuran sama.


**Contoh Kode (Perkalian Hadamard):**

```python
# Dalam NumPy, operator '*' melakukan perkalian Hadamard
C = A * B
```

# **Perkalian Matriks Standar (Standard Matrix Multiplication)**
Berbeda dengan Hadamard, perkalian standar (notasi AB) melibatkan perkalian titik (dot product) antara baris matriks kiri dan kolom matriks kanan. Syarat utamanya adalah dimensi "dalam" harus cocok (M×N dikali N×K menghasilkan matriks M×K).


**Contoh Kode (Perkalian Standar):**

```python
# Menggunakan operator '@' untuk perkalian matriks standar
result = A @ B
```

**Contoh Kode (Transpose):**

```python
# Menggunakan method .T
A_transpose = A.T
```

# **Matriks Simetris (Symmetric Matrices)**
Matriks simetris adalah matriks persegi yang nilainya sama dengan hasil transpose-nya (A=A^T). Properti ini membuat mereka sangat stabil secara numerik dan krusial dalam algoritme komputer. Matriks simetris dapat dibuat dari matriks apa pun (bahkan matriks non-persegi) dengan mengalikannya dengan transpose-nya sendiri (A^T A atau AA^T).


**Contoh Kode (Multiplicative Method):**

```python
# Membuat matriks simetris dari matriks acak A
S = A.T @ A
```
