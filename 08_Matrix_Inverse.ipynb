# **Definisi Inversi Matriks**
Secara matematis, inversi dari matriks A adalah matriks lain A^−1 yang jika dikalikan dengan A akan menghasilkan matriks identitas (I). Dalam aljabar biasa, ini serupa dengan membagi angka (misalnya, inversi dari 3 adalah 1/3 karena 3×1/3=1). Inversi digunakan untuk mengisolasi variabel x dalam persamaan Ax=b menjadi x=A^−1 b.


**Contoh Kode (Menggunakan np.linalg.inv):**

```python
import numpy as np

# Membuat matriks persegi full-rank
A = np.array([[6, 7], [8, 9]])

# Menghitung inversi
A_inv = np.linalg.inv(A)

# Memverifikasi A @ A_inv = I
print(A @ A_inv)
```

# **Jenis-jenis Inversi dan Kondisi Invertibilitas**
Tidak semua matriks dapat dibalik. Hanya matriks persegi yang memiliki peringkat penuh (full-rank) yang memiliki inversi penuh. Matriks yang tidak memiliki inversi disebut matriks singular atau rank-deficient. Jika matriks tersebut singular, Python akan mengeluarkan LinAlgError.

# **Menghitung Inversi: Kasus Khusus**
* **Matriks 2x2**: Terdapat jalan pintas dengan menukar elemen diagonal, mengubah tanda elemen non-diagonal menjadi negatif, dan membaginya dengan determinan.
* **Matriks Diagonal**: Cukup dengan membalik (mengambil resiprokal) dari setiap elemen pada diagonal utamanya. Jika ada satu saja angka nol pada diagonal, maka matriks tersebut tidak dapat dibalik.

**Contoh Kode (Inversi Matriks Diagonal):**

```python
# Matriks diagonal
D = np.diag([7, 8, 14])

# Inversinya cukup dengan 1/elemen diagonal
D_inv = np.diag(1 / np.diag(D))
```

Inversi Satu Sisi (One-Sided Inverses)
Matriks non-persegi (rectangular) tidak memiliki inversi penuh, tetapi mungkin memiliki inversi satu sisi:
* **Left-Inverse**: Berlaku untuk matriks "tinggi" (tall matrix) dengan full column-rank. Digunakan untuk menyelesaikan LT=I.
* **Right-Inverse**: Berlaku untuk matriks "lebar" (wide matrix) dengan full row-rank untuk menyelesaikan WR=I.

**Contoh Kode (Left-Inverse):**

```python
# T adalah matriks tinggi (misal 40x4)
# Formula: L = (T.T @ T)^-1 @ T.T
L = np.linalg.inv(T.T @ T) @ T.T

# L @ T akan menghasilkan Identitas
print(L @ T)
```

# **Moore-Penrose Pseudoinverse**
Setiap matriks, terlepas dari bentuk atau peringkatnya (bahkan yang singular), memiliki pseudoinverse. Pseudoinverse (notasi A†) adalah matriks transformasi yang membawa matriks asal sedekat mungkin ke matriks identitas. Dalam Python, ini dihitung menggunakan fungsi np.linalg.pinv.

**Contoh Kode (Pseudoinverse):**

```python
# Bekerja pada matriks singular sekalipun
A_singular = np.array([[6, 7], [8, 21]]) # Rank-1
A_pinv = np.linalg.pinv(A_singular)

print(A_singular @ A_pinv) # Mendekati Identitas tapi tidak sempurna
```

# **Stabilitas Numerik dan Interpretasi Geometris**
Menghitung inversi pada komputer melibatkan banyak operasi titik mengambang yang rentan terhadap kesalahan pembulatan, terutama pada matriks ill-conditioned (seperti matriks Hilbert). Secara geometris, inversi matriks dapat dianggap sebagai tindakan membatalkan (undoing) transformasi geometris yang telah dilakukan oleh matriks asalnya.
